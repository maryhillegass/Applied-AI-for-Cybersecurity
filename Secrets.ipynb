```python
# Install the Gemini SDK (run once in Colab)
!pip install -q google-generativeai

import google.generativeai as genai
from google.colab import userdata

# Store your API key in Colab's Secrets (key icon in left sidebar) as GEMINI_API_KEY

genai.configure(api_key=userdata.get('Gemini'))

model = genai.GenerativeModel('gemini-3.6-flash')
response = model.generate_content("Explain what an SNS topic is in one sentence.")

print(response.text)
```
