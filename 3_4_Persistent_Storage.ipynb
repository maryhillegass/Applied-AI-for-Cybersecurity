```python
from google.colab import drive
import os

# Mount Google Drive if not already mounted
# If you see 'Drive already mounted at /content/drive; skipping.', it's fine.
try:
    drive.mount('/content/drive')
except Exception as e:
    print(f"Could not mount drive: {e}")
    print("Please ensure you're running this in a Colab environment and authorize access.")

# Define the file path within your Google Drive
drive_path = '/content/drive/MyDrive/'
file_name = 'my_test_file.txt'
full_path = os.path.join(drive_path, file_name)

# Ensure the directory exists (optional, as MyDrive usually exists)
os.makedirs(drive_path, exist_ok=True)

# Content to write to the file
file_content = "Hello from Colab! This file was saved to Google Drive.\n"
file_content += "You can find it in your 'My Drive' folder."

# Write the content to the file
try:
    with open(full_path, 'w') as f:
        f.write(file_content)
    print(f"File '{file_name}' saved successfully to: {full_path}")
except Exception as e:
    print(f"Error saving file: {e}")
    print("Please check if your Google Drive is correctly mounted and you have write permissions.")
```

```python
from google.colab import drive
import os

# Forcibly mount Google Drive
# This is useful after a runtime restart or if the drive is not properly mounted.
try:
    drive.mount('/content/drive', force_remount=True)
    print("Google Drive mounted successfully.")
except Exception as e:
    print(f"Error mounting Google Drive: {e}")
    print("Please ensure you authorize Google Drive access.")

# Define the file path within your Google Drive
drive_path = '/content/drive/MyDrive/'
file_name = 'my_test_file.txt'
full_path = os.path.join(drive_path, file_name)

# Read and print the content of the file
try:
    with open(full_path, 'r') as f:
        content = f.read()
        print(f"\nContents of '{file_name}':")
        print(content)
except FileNotFoundError:
    print(f"Error: The file '{file_name}' was not found at '{full_path}'.")
    print("Please ensure the file exists and the path is correct.")
except Exception as e:
    print(f"An unexpected error occurred while reading the file: {e}")
```
